and execute a code cell with

```
%%hey [problem_file=...] [answer_file=...]

... arbitrary texts ...

```

### 1-1-1. Builtin variables
Following variables are available in `%%hey` cells and replaced with respective strings

* `{file:FILENAME}` = the content of FILE
* `{files:PATTERN}` = the content of files matching PATTERN (e.g., `{files:go/h/*/*.go}`)
* `{bash[-1]}` = the output of the last `%%bash_` cell, `{bash[-2]}` that of the second last `%%bash_` cell, etc.
* `{problem}` = the content of the file you specified by `%%hey problem_file=foo.md`
* `{answer}` = the content of the file you specified by `%%hey answer_file=go/foo.go`


# 2. Go

## 2-2. Source code

````python
%%writefile_ go/h.go

package pl06
````

## 2-3. Generate assembly code

````python
%%bash_

export PATH=${PATH}:~/.local/go/bin:~/go/bin
go build -o go/h.o go/h.go
go tool objdump -gnu go/h.o > go/h.s
cat go/h.s
````
