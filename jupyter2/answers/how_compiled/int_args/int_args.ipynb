# 2. Go

## 2-1. AI tutor

````python
import heytutor
````

## 2-2. Source code

````python
%%writefile_ go/int_args.go

package how_compiled
func IntArgs(a00 int64, a01 int64, a02 int64, a03 int64, a04 int64, a05 int64,
    a06 int64, a07 int64, a08 int64, a09 int64, a10 int64, a11 int64) int64 {
    return a00 + a01 + a02 + a03 + a04 + a05 + a06 + a07 + a08 + a09 + a10 + a11
}
````

## 2-3. Generate assembly code

````python
%%bash_

export PATH=${PATH}:~/.local/go/bin:~/go/bin
go build -o go/int_args.o go/int_args.go
go tool objdump -gnu go/int_args.o > go/int_args.s
cat go/int_args.s
````
