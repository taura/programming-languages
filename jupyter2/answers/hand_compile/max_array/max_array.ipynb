## Problem

* Write a function `max_array(a, n)` that computes the maximum value of an `n`-element array of `double`s `a`, __in assembly__.
* You may assume all elements are positive and return `0` if there are no elements (`n == 0`).
* That is, translate the following C function into assembly:
```
double max_array(double * a, long n) {
  double m = 0.0;
  for (long i = 0; i < n; i++) {
    if (a[i] > m) m = a[i];
  }
  return m;
}
```
* Fill in the skeleton `max_array.s` (after `// ------- write your answer here -------`).

````python
%%writefile_ explore.c
/* A for loop over doubles. Observe the floating-point arithmetic inside the loop
   and how the loop counter (an integer) drives the iteration. */
double dax_b(double x0, double a, double b, long n) {
  double x = x0;
  for (long i = 0; i < n; i++) {
    x = a * x + b;
  }
  return x;
}
````


# 4. Checker
* The following C program calls your `max_array` function and checks the result against a reference computed in C.

````python
%%writefile_ check_max_array.c
#include <assert.h>
#include <stdio.h>
#include <stdlib.h>
double max_array(double *, long);

double max_array_c(double * a, long n) {
  double m = 0.0;
  for (long i = 0; i < n; i++) {
    if (a[i] > m) m = a[i];
  }
  return m;
}

int main(int argc, char ** argv) {
  long n = argc - 1;
  double * a = (double *)malloc(sizeof(double) * n);
  for (long i = 0; i < n; i++) {
    a[i] = atof(argv[i + 1]);
  }
  double ma = max_array(a, n);
  double ma_c = max_array_c(a, n);
  if (ma == ma_c) {
    printf("OK %f %f\n", ma, ma_c);
    return 0;
  } else {
    printf("NG %f %f\n", ma, ma_c);
    return 1;
  }
}
````
